```python
df = pd.read_csv('emotion predic for NLP.txt',sep = ';',header = None,names = ['text','emotion'])  # as csv file seprate by ';'
                                                        #header = None beacuse we do not want that 1st value will be header, so we put manullay header name
```

```python
unique_emotions = df['emotion'].unique()   # To see unique values of emotion
emotion_num = {} ## Now we want to label those unique values
i = 0
for emotion in unique_emotions:
    emotion_num[emotion] = i
    i += 1
df['emotion'] = df['emotion'].map(emotion_num)
df
```

# Step 1 : Lower casing

```python
df['text'] = df['text'].apply(lambda x: x.lower())
```

# Step 2 : Remove punctuation

```python
import string
def remove_punctuation(text):
    for punc in string.punctuation:
        text = text.replace(punc,'')
    return text
df['text'] = df['text'].apply(remove_punctuation)
```

# Step 3 : Remove NUmbers

```python
def remove_numbers(text):
    text = ''.join(word for word in text if not word.isdigit())
    return text
df['text'] = df['text'].apply(remove_numbers)
```

# Step 4 : Remove Url

```python
# There has no url so we can skip the part for the data set
```

# Step 5: Remove Emojis & Special Char

```python
# Every emoji has a unique code or ascy values , so we need to find those and remove
```

```python
def remove_emojis(txt):
  new =""
  for i in txt:
    if i.isascii():
      new+=i
  return new
df['text'] = df['text'].apply(remove_emojis)
```

# Step 7 : Remove Stopwords

```python

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('stopwords')  # Stopwords you know
nltk.download('punkt')  # Use for tokenization
```

```python
stop_words = set(stopwords.words('english')) # Import all stiopwords  # we use more in DL
```

```python
stop_words
```

```python
len(stop_words)  # 198 stopwords total
```

```python
df.loc[1]['text']
```

```python
# from here we need to remove stopwords and convert to an array and list
```

```python
# Then we need to tokenize
```

```python
def remove(txt):
  words = txt.split()  # use  split beacuse word_tokenize is not working here
  new_words = []
  for word in words:
    if word not in stop_words:
      new_words.append(word)
  return ' '.join(new_words)
```

```python
df['text'] = df['text'].apply(remove)
```

```python
df.loc[1]['text']  # Now you can see the sentence after remove stopwords
```

```python
from sklearn.model_selection import train_test_split   # Do train test split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.33, random_state=42)
```

# Convert text to Vector for ML

```python
# Here we use bag of words and TF-IDF ,then we will see what will give best ans
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer,CountVectorizer # Import count.. for bag of vector and tfid .. is for TF_IDF vetorize
```

```python
# Bag or words
```

```python
bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
X_train_bow
```

```python
X_test_bow
```

# Use Naive byes ML model

```python
from sklearn.naive_bayes import MultinomialNB  # You can search which naive byes use for the project , as there has many naive byes as like gaussian
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix  #To see accuracy score
```

```python
nb_model = MultinomialNB()
nb_model.fit(X_train_bow,y_train)
```

```python
y_pred = nb_model.predict(X_test_bow)
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
# TF -IDF
```

```python
tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)
```

```python
               # use nave byas 1st in IF-IDF
```

```python
nb2_model = MultinomialNB()
nb2_model.fit(X_train_tfidf,y_train)
```

```python
y_pred = nb2_model.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```
